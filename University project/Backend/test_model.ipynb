```python
from transformers import pipeline

generator = pipeline(
    "text2text-generation",
    model="google/flan-t5-base"
)

text = """
Photosynthesis is a biochemical process by which plants convert light energy into chemical energy,
primarily in the form of glucose, while releasing oxygen as a byproduct.
"""

prompt = f"""
Rewrite the following text so that a 5th-grade student can understand it.
Keep the meaning correct and use simple words.

Text:
{text}
"""

result = generator(prompt, max_new_tokens=128, do_sample=False)
print(result[0]["generated_text"])
```
